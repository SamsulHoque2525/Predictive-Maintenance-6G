```python
# Rolling window size
WINDOW = 5

# Temperature
df_clean["Temperature_Rolling_Mean"] = (
    df_clean.groupby("Machine_ID")["Temperature_C"]
    .transform(lambda x: x.rolling(WINDOW, min_periods=1).mean())
)

# Vibration
df_clean["Vibration_Rolling_Mean"] = (
    df_clean.groupby("Machine_ID")["Vibration_Hz"]
    .transform(lambda x: x.rolling(WINDOW, min_periods=1).mean())
)

# Power consumption
df_clean["Power_Rolling_Mean"] = (
    df_clean.groupby("Machine_ID")["Power_Consumption_kW"]
    .transform(lambda x: x.rolling(WINDOW, min_periods=1).mean())
)

# Error rate
df_clean["Error_Rate_Rolling_Mean"] = (
    df_clean.groupby("Machine_ID")["Error_Rate_%"]
    .transform(lambda x: x.rolling(WINDOW, min_periods=1).mean())
)

# Production speed
df_clean["Production_Speed_Rolling_Mean"] = (
    df_clean.groupby("Machine_ID")["Production_Speed_units_per_hr"]
    .transform(lambda x: x.rolling(WINDOW, min_periods=1).mean())
)

print("Rolling features created successfully!")
```

```python
efficiency_risk_map = {
    "High": 0,
    "Medium": 50,
    "Low": 100
}

df_clean["Efficiency_Risk_Score"] = (
    df_clean["Efficiency_Status"]
    .map(efficiency_risk_map)
)

print(
    df_clean[
        ["Efficiency_Status", "Efficiency_Risk_Score"]
    ].drop_duplicates()
)
```

```python
temperature_threshold = df_clean["Temperature_C"].quantile(0.95)

df_clean["Temperature_Risk"] = (
    df_clean["Temperature_C"] >= temperature_threshold
).astype(int)

print("Temperature threshold:", temperature_threshold)
```

```python
vibration_threshold = df_clean["Vibration_Hz"].quantile(0.95)

df_clean["Vibration_Risk"] = (
    df_clean["Vibration_Hz"] >= vibration_threshold
).astype(int)

print("Vibration threshold:", vibration_threshold)
```

```python
error_threshold = df_clean["Error_Rate_%"].quantile(0.95)

df_clean["Error_Risk"] = (
    df_clean["Error_Rate_%"] >= error_threshold
).astype(int)

print("Error threshold:", error_threshold)
```

```python
latency_threshold = df_clean["Network_Latency_ms"].quantile(0.95)

df_clean["Latency_Risk"] = (
    df_clean["Network_Latency_ms"] >= latency_threshold
).astype(int)

print("Latency threshold:", latency_threshold)
```

```python
packet_loss_threshold = df_clean["Packet_Loss_%"].quantile(0.95)

df_clean["Packet_Loss_Risk"] = (
    df_clean["Packet_Loss_%"] >= packet_loss_threshold
).astype(int)

print("Packet-loss threshold:", packet_loss_threshold)
```

```python
sensor_risk = (
    df_clean["Temperature_Risk"]
    + df_clean["Vibration_Risk"]
    + df_clean["Error_Risk"]
) / 3 * 100

network_risk = (
    df_clean["Latency_Risk"]
    + df_clean["Packet_Loss_Risk"]
) / 2 * 100

df_clean["Maintenance_Risk_Score"] = (
    0.50 * df_clean["Anomaly_Score"]
    + 0.25 * df_clean["Efficiency_Risk_Score"]
    + 0.15 * sensor_risk
    + 0.10 * network_risk
)

df_clean["Maintenance_Risk_Score"] = (
    df_clean["Maintenance_Risk_Score"]
    .clip(0, 100)
    .round(2)
)

print(
    df_clean[
        [
            "Machine_ID",
            "Anomaly_Score",
            "Efficiency_Risk_Score",
            "Maintenance_Risk_Score"
        ]
    ].head(20)
)
```

```python
def maintenance_priority(score):
    if score < 25:
        return "Normal"
    elif score < 50:
        return "Monitor"
    elif score < 75:
        return "Warning"
    elif score < 90:
        return "High Risk"
    else:
        return "Critical"


df_clean["Maintenance_Priority"] = (
    df_clean["Maintenance_Risk_Score"]
    .apply(maintenance_priority)
)

print(
    df_clean["Maintenance_Priority"]
    .value_counts()
)
```

```python
def maintenance_recommendation(row):

    if row["Maintenance_Priority"] == "Critical":
        return "Immediate inspection recommended"

    elif row["Maintenance_Priority"] == "High Risk":
        return "Schedule maintenance inspection soon"

    elif row["Maintenance_Priority"] == "Warning":
        return "Monitor machine condition closely"

    elif row["Maintenance_Priority"] == "Monitor":
        return "Continue monitoring sensor trends"

    else:
        return "Normal operation"


df_clean["Maintenance_Recommendation"] = (
    df_clean.apply(
        maintenance_recommendation,
        axis=1
    )
)
```
